```python
from detector import Detector
from source import Source
import numpy as np


detector_1 = Detector(2, 40, 0.1, 0.5)
detector_1.place_photomultipliers()

start_point = np.array([0, 0, 0])
source = Source(10**4, start_point)

photon_directions = source.generate_directions(1)
```

```python
cross_points = detector_1.find_cross_point(source.position, photon_directions)
pm_points = detector_1.find_photomultiplier(cross_points)
hits = np.sum(pm_points != -1)
print(f'число попаданий фотонов в ФЭУ: {hits}')
```

```python
detections = detector_1.generate_photo_electrons(hits)
print(f'Полный собранный заряд заряд {detections}')
print(f'Ожидалось: {detector_1.probability_a * source.light_yield * detector_1.photomultiplier * detector_1.photocathode_r**2/detector_1.radius**2}')
print('Видимо, ошибка в условии, забыли про множитель 1/4, из-за площади сферы')
print(f'Ожидалось V2: {detector_1.probability_a * source.light_yield * detector_1.photomultiplier * detector_1.photocathode_r**2/detector_1.radius**2/4}')
```
