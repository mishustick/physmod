```python
from source import Source



source = Source(10**4)
directions = source.generate_directions(1.0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# проверка корректности
def test_norm(dirs):
    norm = np.linalg.norm(dirs, axis=1)
    assert np.allclose(norm, np.ones(len(norm)), atol=1e-10), 'Ошибка нормировки!'
    print('Проверка нормы пройдена')

def test_mean(dirs):
    mean = np.sum(dirs, axis=0)/len(dirs)
    # требуем попадание хотя бы в 5 sigm
    assert np.allclose(mean, 0, atol=5/np.sqrt(3*len(dirs))), 'Ошбика среднего!'
    print('Проверка среднего пройдена')

def histograms(dirs):
    phi = np.arctan2(dirs[:,1], dirs[:,0])

    plt.figure(figsize=(8,6))
    plt.subplot(211)
    plt.hist(dirs[:, 2], bins=50, density=True)
    plt.axhline(1/2, color='k')
    plt.xlabel('cos theta')

    plt.subplot(212)
    plt.hist(phi, bins=50, color='red', density=True)
    plt.axhline(1/(2*np.pi), color='k')
    plt.xlabel('phi')

    plt.show()


```

```python
test_norm(directions)
test_mean(directions)
histograms(directions)
```
