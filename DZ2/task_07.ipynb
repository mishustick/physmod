```python
import matplotlib.pyplot as plt

from detector import Detector
from source import Source
import numpy as np
import time


detector_1 = Detector(2, 40, 0.1, 0.5)
detector_1.place_photomultipliers()
# Я делал сразу в векторных вычислениях, поэтому имитирую использование list с помощью цикла
```

```python
t_start = time.perf_counter()
Q = 0
source_1 = Source(10**4, [0, 0, 0])
photons = source_1.generate_photons(1, 1) # генерация по одному фотону, что эквивалентно использованию list
for photon in photons:
    cross_photons = detector_1.find_cross_point_damp(source_1.position, photon) # считаем точки пересечения, находим долетевшие фотоны
    detectors = detector_1.find_photomultiplier(cross_photons) # определяем детекторы
    hits = np.sum(detectors != -1) # считаем попадания
    photo_electrons = detector_1.generate_photo_electrons(hits) # генерируем фэ

    Q += photo_electrons

print(Q)
print('Время на реализацию:',time.perf_counter() - t_start)
t_1 = time.perf_counter() - t_start
```

```python
t_start = time.perf_counter()
Q = 0
source_1 = Source(10**4, [0, 0, 0])
photons = source_1.generate_photons(1, 10**10) # Генерация сразу всех фотонов
for photon in photons:
    cross_photons = detector_1.find_cross_point_damp(source_1.position, photon) # считаем точки пересечения, находим долетевшие фотоны
    detectors = detector_1.find_photomultiplier(cross_photons) # определяем детекторы
    hits = np.sum(detectors != -1) # считаем попадания
    photo_electrons = detector_1.generate_photo_electrons(hits) # генерируем фэ

    Q += photo_electrons

print(Q)
print('Время на реализацию:',time.perf_counter() - t_start)
t_2 = time.perf_counter() - t_start
```
