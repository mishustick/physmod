```python
from detector import Detector
from source import Source
import numpy as np
import time

detector_1 = Detector(15, 2000, 0.1, 0.5)
detector_1.place_photomultipliers()

source_1 = Source(10**4, [0,0,0])

photons = list(source_1.generate_photons(1, 100)) # будем генерировать пачкой по сто фотонов
```

```python
t1_start = time.perf_counter()

Q = 0 # собранный заряд
n_photons = 0
for k, photon in enumerate(photons):
    points = detector_1.find_cross_point(source_1.position, photon)
    phms = detector_1.find_photomuls_fast(points)
    hits = np.sum(phms != -1)
    photo_electrons = detector_1.generate_photo_electrons(hits)
    Q += photo_electrons
    n_photons += len(photon)

t1_finish = time.perf_counter()
```

```python
print(f'Сгенерировано {n_photons} фотонов')
print(f'Собранный заряд {Q}')
print(f'Время выполнения быстрым способом {t1_finish - t1_start}')
```

```python
t2_start = time.perf_counter()
Q = 0 # собранный заряд
n_photons = 0
for k, photon in enumerate(photons):
    points = detector_1.find_cross_point(source_1.position, photon)
    phms = detector_1.find_photomultiplier(points)
    hits = np.sum(phms != -1)
    photo_electrons = detector_1.generate_photo_electrons(hits)
    Q += photo_electrons
    n_photons += len(photon)

t2_finish = time.perf_counter()
```

```python
print(f'Сгенерировано {n_photons} фотонов')
print(f'Собранный заряд {Q}')
print(f'Время выполнения медленным способом {t2_finish - t2_start}')
print(f't_мед/t_быстр = {(t2_finish - t2_start) / (t1_finish - t1_start)}')
```
